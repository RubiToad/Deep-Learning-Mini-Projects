```python
# ================================
# 1. Imports and Setup
# ================================
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB0
from sklearn.metrics import classification_report

# Set seeds for reproducibility
tf.random.set_seed(123)
np.random.seed(123)

# ================================
# 2. Mount Google Drive and Set Paths
# ================================
from google.colab import drive
drive.mount('/content/drive')

# Based on your file structure: "archive (1)" contains a folder "Data", which holds "images_original"
base_path = '/content/drive/MyDrive/archive (1)/Data'
IMAGES_DIR = os.path.join(base_path, 'images_original')

# Verify the directory structure:
print("Contents of Data folder:", os.listdir(base_path))
print("Contents of images_original:", os.listdir(IMAGES_DIR))

# ================================
# 3. Load the Image Data (Spectrogram PNG Files)
# ================================
# We'll use TensorFlow's image_dataset_from_directory to load the PNG files.
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32

train_image_ds = tf.keras.preprocessing.image_dataset_from_directory(
    IMAGES_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

val_image_ds = tf.keras.preprocessing.image_dataset_from_directory(
    IMAGES_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

# Extract class (genre) names from the data
genres = train_image_ds.class_names
NUM_CLASSES = len(genres)
print("Detected genres (from images):", genres)

# ================================
# 4. Build the Models
# ================================

# ---- Model 1: Basic CNN on Images ----
def build_basic_cnn(input_shape=(224, 224, 3), num_classes=NUM_CLASSES):
    model = models.Sequential([
        layers.Conv2D(32, (3, 3), activation='relu', input_shape=input_shape),
        layers.MaxPooling2D(),
        layers.Conv2D(64, (3, 3), activation='relu'),
        layers.MaxPooling2D(),
        layers.Conv2D(128, (3, 3), activation='relu'),
        layers.MaxPooling2D(),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])
    return model

basic_cnn = build_basic_cnn()
basic_cnn.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
basic_cnn.summary()

# ---- Model 2: Transfer Learning using EfficientNetB0 on Images ----
def build_transfer_model(input_shape=(224, 224, 3), num_classes=NUM_CLASSES):
    base_model = EfficientNetB0(weights='imagenet', include_top=False, input_shape=input_shape)
    base_model.trainable = False  # Freeze the pre-trained layers
    inputs = layers.Input(shape=input_shape)
    x = base_model(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)
    model = models.Model(inputs, outputs)
    return model

transfer_model = build_transfer_model()
transfer_model.compile(optimizer='adam',
                       loss='sparse_categorical_crossentropy',
                       metrics=['accuracy'])
transfer_model.summary()

# ================================
# 5. Train the Models
# ================================
EPOCHS = 10  # Adjust the number of epochs as needed

print("\n--- Training Basic CNN on Images ---")
history_basic = basic_cnn.fit(
    train_image_ds,
    validation_data=val_image_ds,
    epochs=EPOCHS
)

print("\n--- Training Transfer Learning Model on Images ---")
history_transfer = transfer_model.fit(
    train_image_ds,
    validation_data=val_image_ds,
    epochs=EPOCHS
)

# ================================
# 6. Evaluate the Models: Classification Reports
# ================================
def get_classification_report(model, dataset, model_name="Model"):
    y_true = []
    y_pred = []
    for batch in dataset:
        images, labels = batch
        preds = model.predict(images)
        y_pred.extend(np.argmax(preds, axis=1))
        y_true.extend(labels.numpy())
    report = classification_report(y_true, y_pred, target_names=genres)
    print(f"\nClassification Report for {model_name}:\n{report}")

# Evaluate models on validation data
get_classification_report(basic_cnn, val_image_ds, model_name="Basic CNN")
get_classification_report(transfer_model, val_image_ds, model_name="Transfer Learning Model")

# ================================
# 7. Plot Training & Validation Curves
# ================================
def plot_history(history, model_name="Model"):
    plt.figure(figsize=(12, 5))

    # Plot Accuracy vs. Epoch
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history['val_accuracy'], label='Val Accuracy')
    plt.title(f'{model_name} Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    # Plot Loss vs. Epoch
    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Val Loss')
    plt.title(f'{model_name} Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.show()

plot_history(history_basic, model_name="Basic CNN")
plot_history(history_transfer, model_name="Transfer Learning")
```

```python
# ================================
# 1. Imports and Setup
# ================================
import os
import numpy as np
import tensorflow as tf
import librosa
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB0
from sklearn.metrics import classification_report

# Set seeds for reproducibility
tf.random.set_seed(123)
np.random.seed(123)

# ================================
# 2. Mount Google Drive and Set Paths
# ================================
from google.colab import drive
drive.mount('/content/drive')

# Your file structure:
# "/content/drive/MyDrive/archive (1)/Data/" contains:
#   - images_original (PNG files, sorted in subfolders by genre)
#   - genres_original (WAV files, sorted in subfolders by genre)
base_path = '/content/drive/MyDrive/archive (1)/Data'
IMAGES_DIR = os.path.join(base_path, 'images_original')
AUDIO_DIR = os.path.join(base_path, 'genres_original')

# Verify the directory structure:
print("Contents of Data folder:", os.listdir(base_path))
print("Contents of images_original:", os.listdir(IMAGES_DIR))
print("Contents of genres_original:", os.listdir(AUDIO_DIR))

# ================================
# 3. Load the Image Data (PNG Spectrograms)
# ================================
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32

train_image_ds = tf.keras.preprocessing.image_dataset_from_directory(
    IMAGES_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

val_image_ds = tf.keras.preprocessing.image_dataset_from_directory(
    IMAGES_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

# The folder names are our genre labels.
genres = train_image_ds.class_names
NUM_CLASSES = len(genres)
print("Detected genres (from images):", genres)

# ================================
# 4. Build the Audio Dataset (WAV Files) with Pre-Filtering
# ================================

# Step 4a: Collect all file paths and labels from the audio folder
audio_file_paths = []
audio_labels = []
for genre in sorted(os.listdir(AUDIO_DIR)):
    genre_folder = os.path.join(AUDIO_DIR, genre)
    if os.path.isdir(genre_folder):
        for file in os.listdir(genre_folder):
            if file.endswith('.wav'):
                audio_file_paths.append(os.path.join(genre_folder, file))
                audio_labels.append(genre)

# Map genre names to numeric labels using the same ordering as in images.
label_to_index = {name: idx for idx, name in enumerate(genres)}
audio_labels_idx = [label_to_index[label] for label in audio_labels if label in label_to_index]

print("Total audio files before filtering:", len(audio_file_paths))

# Step 4b: Pre-filter audio files by attempting to load a short segment.
# Only keep files that can be loaded successfully.
valid_audio_file_paths = []
valid_audio_labels_idx = []
for fp, lbl in zip(audio_file_paths, audio_labels_idx):
    try:
        # Try loading just 0.5 seconds to check format without much delay.
        librosa.load(fp, sr=22050, duration=0.5)
        valid_audio_file_paths.append(fp)
        valid_audio_labels_idx.append(lbl)
    except Exception as e:
        print(f"Skipping file {fp} due to error: {e}")

print("Total valid audio files:", len(valid_audio_file_paths))

# Step 4c: Define the function to load a full audio file and compute its log-mel spectrogram.
def load_audio_melspectrogram(file_path, sr=22050, n_mels=128, n_fft=2048, hop_length=512, duration=3):
    y, sr = librosa.load(file_path, sr=sr, duration=duration)
    S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels, n_fft=n_fft, hop_length=hop_length)
    log_S = librosa.power_to_db(S, ref=np.max)
    # Force fixed number of frames (128); pad if needed or truncate.
    desired_frames = 128
    if log_S.shape[1] < desired_frames:
        pad_width = desired_frames - log_S.shape[1]
        log_S = np.pad(log_S, ((0, 0), (0, pad_width)), mode='constant')
    else:
        log_S = log_S[:, :desired_frames]
    log_S = np.expand_dims(log_S, axis=-1)  # Shape: (128, 128, 1)
    return log_S.astype(np.float32)

# Step 4d: Create a generator for the audio dataset.
def audio_generator(file_paths, labels):
    for fp, lbl in zip(file_paths, labels):
        try:
            features = load_audio_melspectrogram(fp)
            yield features, lbl
        except Exception as e:
            # This should rarely occur because we pre-filtered, but just in case.
            print(f"Error processing {fp}: {e}")
            continue

audio_ds = tf.data.Dataset.from_generator(
    lambda: audio_generator(valid_audio_file_paths, valid_audio_labels_idx),
    output_signature=(
        tf.TensorSpec(shape=(128, 128, 1), dtype=tf.float32),
        tf.TensorSpec(shape=(), dtype=tf.int32)
    )
)

# Shuffle and split the audio dataset (80% train, 20% validation)
audio_count = len(valid_audio_file_paths)
audio_ds = audio_ds.shuffle(buffer_size=audio_count, seed=123)
val_audio_ds = audio_ds.take(int(0.2 * audio_count))
train_audio_ds = audio_ds.skip(int(0.2 * audio_count))
train_audio_ds = train_audio_ds.batch(BATCH_SIZE)
val_audio_ds = val_audio_ds.batch(BATCH_SIZE)

# ================================
# 5. Build the Multimodal Model (Image + Audio)
# ================================
def build_multimodal_model(image_input_shape=(224, 224, 3), audio_input_shape=(128, 128, 1), num_classes=NUM_CLASSES):
    # Image branch
    image_input = layers.Input(shape=image_input_shape, name='image_input')
    x1 = layers.Conv2D(32, (3,3), activation='relu')(image_input)
    x1 = layers.MaxPooling2D()(x1)
    x1 = layers.Conv2D(64, (3,3), activation='relu')(x1)
    x1 = layers.MaxPooling2D()(x1)
    x1 = layers.Flatten()(x1)
    x1 = layers.Dense(128, activation='relu')(x1)

    # Audio branch
    audio_input = layers.Input(shape=audio_input_shape, name='audio_input')
    x2 = layers.Conv2D(16, (3,3), activation='relu')(audio_input)
    x2 = layers.MaxPooling2D()(x2)
    x2 = layers.Conv2D(32, (3,3), activation='relu')(x2)
    x2 = layers.MaxPooling2D()(x2)
    x2 = layers.Flatten()(x2)
    x2 = layers.Dense(64, activation='relu')(x2)

    # Fuse branches
    combined = layers.concatenate([x1, x2])
    x = layers.Dense(128, activation='relu')(combined)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = models.Model(inputs=[image_input, audio_input], outputs=outputs)
    return model

multimodal_model = build_multimodal_model()
multimodal_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
multimodal_model.summary()

# ================================
# 6. Prepare the Combined Dataset for Multimodal Input
# ================================
# IMPORTANT: This code assumes that the ordering of samples in the image dataset
# and the (filtered) audio dataset is the same.
# In many cases, explicit alignment via filenames is preferable.

# Unbatch the training datasets.
image_ds_unbatched = train_image_ds.unbatch()
audio_ds_unbatched = train_audio_ds.unbatch()

# Zip them together: each element becomes ((image, label_image), (audio, label_audio)).
combined_train_ds = tf.data.Dataset.zip((image_ds_unbatched, audio_ds_unbatched))

# Define a function to combine modalities.
def combine_modalities(img_data, audio_data):
    (img, lab_img) = img_data
    (aud, lab_aud) = audio_data
    # We assume lab_img and lab_aud are the same.
    return {"image_input": img, "audio_input": aud}, lab_img

combined_train_ds = combined_train_ds.map(combine_modalities).batch(BATCH_SIZE)

# Prepare the combined validation dataset similarly.
image_val_unbatched = val_image_ds.unbatch()
audio_val_unbatched = val_audio_ds.unbatch()
combined_val_ds = tf.data.Dataset.zip((image_val_unbatched, audio_val_unbatched))
combined_val_ds = combined_val_ds.map(combine_modalities).batch(BATCH_SIZE)

# ================================
# 7. Train the Multimodal Model
# ================================
EPOCHS = 10  # Adjust as needed
print("\n--- Training Multimodal Model (Image + Audio) ---")
history_multimodal = multimodal_model.fit(
    combined_train_ds,
    validation_data=combined_val_ds,
    epochs=EPOCHS
)

# ================================
# 8. Evaluate the Multimodal Model
# ================================
def get_classification_report_multimodal(model, dataset, model_name="Multimodal Model"):
    y_true = []
    y_pred = []
    for batch in dataset:
        inputs, labels = batch
        preds = model.predict(inputs)
        y_pred.extend(np.argmax(preds, axis=1))
        y_true.extend(labels.numpy())
    report = classification_report(y_true, y_pred, target_names=genres)
    print(f"\nClassification Report for {model_name}:\n{report}")

get_classification_report_multimodal(multimodal_model, combined_val_ds, model_name="Multimodal Model")

# ================================
# 9. Plot Training & Validation Curves
# ================================
def plot_history(history, model_name="Multimodal Model"):
    plt.figure(figsize=(12, 5))
    # Accuracy vs Epoch
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history['val_accuracy'], label='Val Accuracy')
    plt.title(f'{model_name} Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    # Loss vs Epoch
    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Val Loss')
    plt.title(f'{model_name} Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.show()

plot_history(history_multimodal, model_name="Multimodal Model")
```

```python
# ================================
# 1. Imports and Setup
# ================================
import os
import numpy as np
import tensorflow as tf
import librosa
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB0
from sklearn.metrics import classification_report

# Set seeds for reproducibility
tf.random.set_seed(123)
np.random.seed(123)

# ================================
# 2. Mount Google Drive and Set Paths
# ================================
from google.colab import drive
drive.mount('/content/drive')

# File structure:
# "/content/drive/MyDrive/archive (1)/Data/" contains:
#   - images_original (PNG files, sorted in subfolders by genre)
#   - genres_original (WAV files, sorted in subfolders by genre)
base_path = '/content/drive/MyDrive/archive (1)/Data'
IMAGES_DIR = os.path.join(base_path, 'images_original')
AUDIO_DIR = os.path.join(base_path, 'genres_original')

# Verify the directory structure:
print("Contents of Data folder:", os.listdir(base_path))
print("Contents of images_original:", os.listdir(IMAGES_DIR))
print("Contents of genres_original:", os.listdir(AUDIO_DIR))

# ================================
# 3. Load the Image Data (PNG Spectrograms)
# ================================
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32

train_image_ds = tf.keras.preprocessing.image_dataset_from_directory(
    IMAGES_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

val_image_ds = tf.keras.preprocessing.image_dataset_from_directory(
    IMAGES_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

# The folder names become our genre labels.
genres = train_image_ds.class_names
NUM_CLASSES = len(genres)
print("Detected genres (from images):", genres)

# ================================
# 4. Build the Audio Dataset (WAV Files) with Pre-Filtering
# ================================

# Collect all file paths and labels
audio_file_paths = []
audio_labels = []
for genre in sorted(os.listdir(AUDIO_DIR)):
    genre_folder = os.path.join(AUDIO_DIR, genre)
    if os.path.isdir(genre_folder):
        for file in os.listdir(genre_folder):
            if file.endswith('.wav'):
                audio_file_paths.append(os.path.join(genre_folder, file))
                audio_labels.append(genre)

# Map genre names to numeric labels (with the same order as in images)
label_to_index = {name: idx for idx, name in enumerate(genres)}
audio_labels_idx = [label_to_index[label] for label in audio_labels if label in label_to_index]
print("Total audio files before filtering:", len(audio_file_paths))

# Pre-filter audio files by trying to load a short segment
valid_audio_file_paths = []
valid_audio_labels_idx = []
for fp, lbl in zip(audio_file_paths, audio_labels_idx):
    try:
        # Attempt to load a short segment (0.5 sec) to ensure the file is readable
        librosa.load(fp, sr=22050, duration=0.5)
        valid_audio_file_paths.append(fp)
        valid_audio_labels_idx.append(lbl)
    except Exception as e:
        print(f"Skipping file {fp} due to error: {e}")

print("Total valid audio files:", len(valid_audio_file_paths))

# Define a function to load a full audio file and compute its log-mel spectrogram.
def load_audio_melspectrogram(file_path, sr=22050, n_mels=128, n_fft=2048, hop_length=512, duration=3):
    y, sr = librosa.load(file_path, sr=sr, duration=duration)
    S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels, n_fft=n_fft, hop_length=hop_length)
    log_S = librosa.power_to_db(S, ref=np.max)
    # Force fixed number of time frames (128); pad or truncate if necessary.
    desired_frames = 128
    if log_S.shape[1] < desired_frames:
        pad_width = desired_frames - log_S.shape[1]
        log_S = np.pad(log_S, ((0, 0), (0, pad_width)), mode='constant')
    else:
        log_S = log_S[:, :desired_frames]
    log_S = np.expand_dims(log_S, axis=-1)  # Shape becomes (128, 128, 1)
    return log_S.astype(np.float32)

# Create a generator for the audio dataset.
def audio_generator(file_paths, labels):
    for fp, lbl in zip(file_paths, labels):
        try:
            features = load_audio_melspectrogram(fp)
            yield features, lbl
        except Exception as e:
            print(f"Error processing {fp}: {e}")
            continue

audio_ds = tf.data.Dataset.from_generator(
    lambda: audio_generator(valid_audio_file_paths, valid_audio_labels_idx),
    output_signature=(
        tf.TensorSpec(shape=(128, 128, 1), dtype=tf.float32),
        tf.TensorSpec(shape=(), dtype=tf.int32)
    )
)

# Shuffle and split the audio dataset (80% train, 20% validation)
audio_count = len(valid_audio_file_paths)
audio_ds = audio_ds.shuffle(buffer_size=audio_count, seed=123)
val_audio_ds = audio_ds.take(int(0.2 * audio_count))
train_audio_ds = audio_ds.skip(int(0.2 * audio_count))
train_audio_ds = train_audio_ds.batch(BATCH_SIZE)
val_audio_ds = val_audio_ds.batch(BATCH_SIZE)

# ================================
# 5. Build the Multimodal Model (Image + Audio)
# ================================
def build_multimodal_model(image_input_shape=(224, 224, 3), audio_input_shape=(128, 128, 1), num_classes=NUM_CLASSES):
    # Image branch
    image_input = layers.Input(shape=image_input_shape, name='image_input')
    x1 = layers.Conv2D(32, (3,3), activation='relu')(image_input)
    x1 = layers.MaxPooling2D()(x1)
    x1 = layers.Conv2D(64, (3,3), activation='relu')(x1)
    x1 = layers.MaxPooling2D()(x1)
    x1 = layers.Flatten()(x1)
    x1 = layers.Dense(128, activation='relu')(x1)

    # Audio branch
    audio_input = layers.Input(shape=audio_input_shape, name='audio_input')
    x2 = layers.Conv2D(16, (3,3), activation='relu')(audio_input)
    x2 = layers.MaxPooling2D()(x2)
    x2 = layers.Conv2D(32, (3,3), activation='relu')(x2)
    x2 = layers.MaxPooling2D()(x2)
    x2 = layers.Flatten()(x2)
    x2 = layers.Dense(64, activation='relu')(x2)

    # Fuse branches
    combined = layers.concatenate([x1, x2])
    x = layers.Dense(128, activation='relu')(combined)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = models.Model(inputs=[image_input, audio_input], outputs=outputs)
    return model

multimodal_model = build_multimodal_model()
multimodal_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
multimodal_model.summary()

# ================================
# 6. Prepare the Combined Dataset for Multimodal Input
# ================================
# Note: This code assumes that the ordering of samples in the image and audio training datasets is identical.
# (If not, you'll need to align samples based on filenames or another mapping method.)

# Unbatch the training datasets to get individual samples.
image_ds_unbatched = train_image_ds.unbatch()
audio_ds_unbatched = train_audio_ds.unbatch()

# Zip the unbatched datasets; each element is a tuple: ((image, label_image), (audio, label_audio))
combined_train_ds = tf.data.Dataset.zip((image_ds_unbatched, audio_ds_unbatched))

# Define a function to combine the two modalities.
def combine_modalities(img_data, audio_data):
    (img, lab_img) = img_data
    (aud, lab_aud) = audio_data
    return {"image_input": img, "audio_input": aud}, lab_img

combined_train_ds = combined_train_ds.map(combine_modalities).batch(BATCH_SIZE)

# Do the same for the validation data.
image_val_unbatched = val_image_ds.unbatch()
audio_val_unbatched = val_audio_ds.unbatch()
combined_val_ds = tf.data.Dataset.zip((image_val_unbatched, audio_val_unbatched))
combined_val_ds = combined_val_ds.map(combine_modalities).batch(BATCH_SIZE)

# --- Repeat the training dataset to avoid running out of data ---
# Also, compute steps_per_epoch based on the number of individual image samples.
total_train_samples = sum(1 for _ in train_image_ds.unbatch())
steps_per_epoch = int(np.ceil(total_train_samples / BATCH_SIZE))
print("Total training samples:", total_train_samples, "Steps per epoch:", steps_per_epoch)

# Repeat the combined training dataset indefinitely.
combined_train_ds = combined_train_ds.repeat()

# ================================
# 7. Train the Multimodal Model
# ================================
EPOCHS = 10  # Adjust as needed
print("\n--- Training Multimodal Model (Image + Audio) ---")
history_multimodal = multimodal_model.fit(
    combined_train_ds,
    steps_per_epoch=steps_per_epoch,
    validation_data=combined_val_ds,
    epochs=EPOCHS
)

# ================================
# 8. Evaluate the Multimodal Model
# ================================
def get_classification_report_multimodal(model, dataset, model_name="Multimodal Model"):
    y_true = []
    y_pred = []
    for batch in dataset:
        inputs, labels = batch
        preds = model.predict(inputs)
        y_pred.extend(np.argmax(preds, axis=1))
        y_true.extend(labels.numpy())
    report = classification_report(y_true, y_pred, target_names=genres)
    print(f"\nClassification Report for {model_name}:\n{report}")

get_classification_report_multimodal(multimodal_model, combined_val_ds, model_name="Multimodal Model")

# ================================
# 9. Plot Training & Validation Curves
# ================================
def plot_history(history, model_name="Multimodal Model"):
    plt.figure(figsize=(12, 5))

    # Accuracy vs. Epoch plot
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history['val_accuracy'], label='Val Accuracy')
    plt.title(f'{model_name} Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    # Loss vs. Epoch plot
    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Val Loss')
    plt.title(f'{model_name} Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.show()

plot_history(history_multimodal, model_name="Multimodal Model")
```
